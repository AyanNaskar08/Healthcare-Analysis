```python
df = df.drop_duplicates()  #df.duplicated().sum() to verify
```

```python
df.loc[df['Billing Amount'] < 0 , 'Billing Amount'] = df['Billing Amount'].mean()   #df.describe() to verify
```

```python
df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
```

```python
df['Discharge Date'] = pd.to_datetime(df['Discharge Date']) #df.dtypes to verify
```

```python
df['Length_of_stay'] = (df['Discharge Date'] - df['Date of Admission']).dt.days
```

## Data Analysis and Data Visualization

## Qusetion 1 : Which medical condition affects te most patients across the network ?

```python
conditional_count = df['Medical Condition'].value_counts()
conditional_count
```

```python
plt.figure(figsize = (8,5))
plt.bar(conditional_count.index, conditional_count.values,color = '#7833fe')
plt.title('Patients by Medical Condition')
plt.xlabel('Medical Condition')
plt.ylabel('Number of patients')
plt.xticks(rotation = 90)
plt.tight_layout()
plt.show()
```

## Question 2 : How is patient age distributed across the network ?

```python
average_age = df['Age'].mean()
average_age
```

```python
plt.figure(figsize= (8,5))
plt.hist(df['Age'], bins = 20 , color = '#ff5722' , edgecolor = 'black' )
plt.title('Age Distribution of patients')
plt.xlabel('Age')
plt.ylabel('Number of patients')
plt.tight_layout()
plt.show()
```

## Question 3 : How has average treatment cost changed overtime ?

```python
average_cost = df['Billing Amount'].mean()
average_cost
```

```python
df['Admission_month'] = df['Date of Admission'].dt.to_period('M')
monthly_average_cost = df.groupby('Admission_month')['Billing Amount'].mean()
monthly_average_cost
```

```python
plt.figure(figsize= (10,5))
plt.plot(monthly_average_cost.index.astype(str),monthly_average_cost.values,color = '#28a745' , marker = 'o' , markersize = 3 )
plt.title('Monthly Average Cost of Patients')
plt.xlabel('Month')
plt.ylabel('Average Billing Amount')
plt.xticks(rotation = 90)
plt.tight_layout()
plt.show()
```

## Question 4 : Is there a releationship between paitent age and treatment cost ?

```python
senior_avg_cost = df.loc[df['Age'] > 60 , 'Billing Amount'].mean()
senior_avg_cost
```

```python
sample_df = df.sample(1000 , random_state = 1)
plt.figure(figsize= (8,5))
plt.scatter(sample_df['Age'], sample_df['Billing Amount'] , color = 'blue' , alpha = 0.5 , s = 15)
plt.title('Age VS Treatment Cost')
plt.xlabel('Age')
plt.ylabel('Billing Amount')
plt.tight_layout()
plt.show()
```

## Question 5 : Which insurance covers the most paitents ?

```python
insurance_provider_count = df['Insurance Provider'].value_counts()
insurance_provider_count
```

```python
plt.figure(figsize = (8,5))
plt.pie(insurance_provider_count.values , labels= insurance_provider_count.index , autopct= '%1.1f%%'  , colors= ['#93b5e1','#848ccf','#be5683','#1dd3bd','#fa7d09'])
plt.title('Insurance Provider Distribution')
plt.tight_layout()
plt.show()
```

Bussiness scenario II : "Hospital fiance term wants to identify patients who have unusually high billing amount so they can review those cause "
